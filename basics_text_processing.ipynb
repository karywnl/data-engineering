```python
import re
```

```python
s1 = 'python is easy to learn and code'
```

```python
print(re.match('easy', s1))
```

character by character match.

```python
print(re.match('Python', s1))   # case sensisitive.
```

```python
print(re.search('Python', s1, re.I))    # telling re.I says it not to be case sensitive
```

```python
print(re.findall('a', s1, re.I))
```

```python
re.split(" ", s1)
```

```python
re.sub(" ", "-", s1)
```

```python
s = 'hello@123'
```

```python
pattern = re.compile('[^A-Za-z]')
re.sub(pattern, '', s)
```
