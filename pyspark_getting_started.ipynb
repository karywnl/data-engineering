# Getting started with PySpark

Run the cells from top to bottom with the **data-engineering** kernel. PySpark is the Python interface; Spark does the processing in a Java process on this Mac.

```python
import os
import sys

os.environ["JAVA_HOME"] = "/opt/homebrew/opt/openjdk@21/libexec/openjdk.jdk/Contents/Home"
os.environ["PYSPARK_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = (SparkSession.builder
         .master("local[2]")
         .appName("DataEngineeringPractice")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
print(f"Spark {spark.version}; Python {sys.version.split()[0]}")
```

```python
rows = [("books", 2), ("books", 3), ("pens", 4)]
df = spark.createDataFrame(rows, ["item", "quantity"])
df.show()
```

```python
from pyspark.sql import functions as F

df.groupBy("item").agg(F.sum("quantity").alias("total")).show()
```

Spark stays available for later cells in this notebook. When finished, run `spark.stop()` or restart the kernel.
